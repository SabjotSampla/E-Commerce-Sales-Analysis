# **Basic Sales** **Statistics**

```python
#Total Sales
total_sales = df["Total_Sales"].sum()
print("Total Sales:",total_sales)
```

```python
# Average Sales
avg_sales = df["Total_Sales"].mean()
print("Average Sales:",avg_sales)
```

```python
#Highest Sales
highest_sales = df["Total_Sales"].max()
print("highest sales:",highest_sales)
```

```python
# Lowest Sales
lowest_sales = df["Total_Sales"].min()
print("Lowest Sales:",lowest_sales)
```

```python
# Total Quantity Sold
total_quantity = df["Quantity"].sum()
print("Total Quantity Sold",total_quantity)
```

```python
# Numbers of Orders
numbers_of_orders = df["Order_ID"].nunique()
print("Numbers of Orders:",numbers_of_orders)
```

# **GroupBy Analysis**

```python
# Sales by Category
category_sales = df.groupby("Category")["Total_Sales"].sum()
category_sales
```

```python
# Average Sales by Category
df.groupby("Category")["Total_Sales"].mean().sort_values(ascending=False)
```

```python
# Quantity Sold by Category
df.groupby("Category")["Order_ID"].count().sort_values(ascending=False)
```

```python
# Number of Orders by Category
df.groupby("Category")["Order_ID"].count().sort_values(ascending = False)
```

# **Aggregate Functions**

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Highest_Sales=("Total_Sales", "max"),
    Lowest_Sales=("Total_Sales", "min"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count"))
category_analysis.sort_values("Total_Sales", ascending=False)
```

# **City Analysis**

```python
# Total Sales by City
city_sales = df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False)
print("City Sales:",city_sales)
```

```python
# Average Sales by City
Average_sales = df.groupby("City")["Total_Sales"].mean().sort_values(ascending=False)
print("Average Sales by City:",Average_sales)
```

```python
# Quantity Sold by City
df.groupby("City")["Quantity"].sum().sort_values(ascending=False)
```

# **Product Analysis**

```python
# Total Sales by Product
product_sales = df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
product_sales
```

```python
# Top 10 Products by Sales
top_products = (
    df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False).head(10))
top_products
```

```python
# Products With Highest Quantity Sold
df.groupby("Product")["Quantity"].sum().sort_values(ascending=False).head(10)
```

# **Payment Method Analysis**

```python
# Number of Orders by Payment Method
df["Payment_Method"].value_counts()
```

```python
# Total Sales by Payment Method
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()
payment_sales.sort_values(ascending=False)
```

```python
# Average Sales by Payment Method
df.groupby("Payment_Method")["Total_Sales"].mean().sort_values(ascending=False)
```

## Meaningful Observations

1. The dataset contains **100 orders** with a total sales value of **₹447,262.95**.

2. The average sales value per order is approximately **₹4,472.63**, while the highest individual order generated **₹13,995** in sales.

3. **Sports** is the highest-performing category with total sales of approximately **₹136,026.45**, followed by **Home & Kitchen** with approximately **₹113,329.90**.

4. **Hyderabad** generated the highest total sales among the cities, with approximately **₹59,076.45**.

5. **Coffee Maker** is the top-performing product by total sales, generating approximately **₹51,610.25**.

6. **UPI** is the most frequently used payment method, accounting for **26 orders**, while Cash on Delivery has the fewest orders.

7. The dataset records a total quantity of **294 products sold** across the 100 orders.

8. The analysis shows that grouping and aggregation by category, city, product, and payment method helps identify the strongest-performing areas of the e-commerce business.
